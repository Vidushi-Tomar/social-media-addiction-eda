# DATA INSIGHTS


**Age Distribution**

```python
sns.set_theme(style="whitegrid", context="notebook")
plt.figure(figsize=(8,5))
ax = sns.countplot(data=df,x='Age',color='#95b5c1',edgecolor='black',linewidth=1.2)
for container in ax.containers:
    ax.bar_label(container, fontsize=10, fontweight='bold', padding=3)
plt.title('Age Distribution', fontsize=16, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Number of Students')
```

**INTERPRETATION**:
Most Number of Students are of Age 20, followed by Age 19 and Age 21.

# EXPLORATORY DATA ANALYSIS

**Comparing Addiction Scores with respect to Gender :**

```python
a=df.groupby(['Gender'],as_index=False)['Addicted_Score'].sum().sort_values(by='Addicted_Score',ascending=False)
sns.set_theme(style="whitegrid",context="notebook",palette="pastel")
ax=sns.barplot(x='Gender',y='Addicted_Score',data=a,color='#d1cb8f',edgecolor='black')
for container in ax.containers:
    ax.bar_label(container, label_type='edge', padding=3)
```

**INTERPRETATION**:
Female Students are more Addicted to Social Media in comparison to Male Students, though the difference is Negligible.

**Comparing Addiction Scores with respect to Age :**

```python
a=df.groupby(['Age'],as_index=False)['Addicted_Score'].sum().sort_values(by='Addicted_Score',ascending=False)
ax=sns.barplot(x='Age',y='Addicted_Score',data=a,color='#74be77',edgecolor='black')
for container in ax.containers:
    ax.bar_label(container, label_type='edge', padding=3)
```

**INTERPRETATION**:
19 Year old Students shows maximum addiction to social media but as we move further till age 24, the addiction scores decreases.

**Comparing Addiction Scores with respect to Academic Level of Students :**

```python
a=df.groupby(['Academic_Level'],as_index=False)['Addicted_Score'].sum().sort_values(by='Addicted_Score',ascending=False)
ax=sns.barplot(x='Academic_Level',y='Addicted_Score',data=a,color='#ffb6c1',edgecolor='black')
for container in ax.containers:
    ax.bar_label(container, label_type='edge', padding=3)
```

**INTERPRETATION**:
Undergraduate Students are more addicted to Social Media than Graduate Students, whereas High School Students' Addiction Score is very low. 

**Comparing Addiction Scores in top 15 countries with most addiction scores:**

```python
a=df.groupby(['Country'],as_index=False)['Addicted_Score'].sum().sort_values(by='Addicted_Score',ascending=False).head(15)
sns.set(rc={'figure.figsize':(17,5)})
ax=sns.barplot(x='Country',y='Addicted_Score',data=a,color='#fbb942',edgecolor='black')
plt.title('Top 15 Countries by Total Addicted Score',size=15) 
for container in ax.containers:
    ax.bar_label(container, label_type='edge', padding=3)
```

**INTERPRETATION**:
Clearly, India Shows the maximum Addiction score of 391 followed by USA, Canada etc.

**To Check which is the Most Used Platform using a Pie Chart:**

```python
platform = df['Most_Used_Platform'].value_counts()
plt.figure(figsize=(8,8))
plt.pie(platform,startangle=90,autopct=lambda p: f'{p:.1f}%' if p >= 5 else '')
plt.legend(platform.index, loc='center left', bbox_to_anchor=(1,0.5))
```

**INTERPRETATION**:
Instagram is the most used platform by Students with a composition of 36%. it is then followed by Tik tok, Facebook and Whatsapp.

**To Check if there is any Correlation between The number of sleep hours and the addiction scores of Students :**

```python
corr = df['Sleep_Hours_Per_Night'].corr(df['Addicted_Score'])
sns.set(rc={'figure.figsize':(8,5)})
sns.regplot(data=df,x='Sleep_Hours_Per_Night',y='Addicted_Score',scatter_kws={'color':'#628F42'},line_kws={'color':'red'})
plt.title(f'Sleep Hours vs Addiction Score (r = {corr:.2f})')
plt.show()
```

**INTERPRETATION**:
Sleep Hours and Addiction Score shows a strong negative correlation between them, i.e. as the number of sleep hours decreases, addiction scores tend to increase among Students.

**To Check if there is any Correlation between The Mental Health Scores and the addiction scores of Students :**

```python
corr = df['Mental_Health_Score'].corr(df['Addicted_Score'])
sns.regplot(data=df,x='Mental_Health_Score',y='Addicted_Score',scatter_kws={'color':'#8F425A'},line_kws={'color':'#EBCA42'})
plt.title(f'Mental Health vs Addiction Score (r = {corr:.2f})')
plt.show()
```

**INTERPRETATION**:
Mental Health and Addiction Score shows a very strong negative correlation between them, i.e. Students with poorer mental Health have more addiction scores. 

**To Check if there is any Correlation between The Mental Health Scores and the involvement in conflicts occuring over social media :**

```python
corr = df['Mental_Health_Score'].corr(df['Conflicts_Over_Social_Media'])
sns.regplot(data=df,x='Mental_Health_Score',y='Conflicts_Over_Social_Media',scatter_kws={'color':'#4291EB'},line_kws={'color':'#EB428E'})
plt.title(f'Mental Health vs Conflicts over Social Media (r = {corr:.2f})')
plt.show()
```

**INTERPRETATION**:
Mental Health and Conflicts over Social Media shows a very strong negative correlation between them, i.e. Students with poorer mental Health tend to have more tendency to indulge in conflicts happenning over Social Media.

**To Check if there is any Correlation between The number of sleep hours and the Mental Health scores of Students :**

```python
sns.boxplot(data=df,x='Mental_Health_Score',y='Sleep_Hours_Per_Night',color='#42B865')
plt.title('Sleep Hours by Mental Health Score')
plt.show()
```

**INTERPRETATION**:
Sleep hours rise steadily with mental health score. students getting 4-5 hours of sleep cluster at low mental health scores (4-5), while those getting 8-9 hrs cluster at high scores (7-8).

**Comparing Most Used Platform with Whether Social Media Affects Academic Performance (Yes/No) :**

```python
plt.figure(figsize=(8,5))
ax = sns.countplot(data=df,x='Most_Used_Platform',hue='Affects_Academic_Performance',palette=['#DE85A0', '#628F42'])
plt.title('Platform vs Academic Performance')
plt.xlabel('Platform')
plt.ylabel('Number of Students')
plt.xticks(rotation=20)
for container in ax.containers:
    ax.bar_label(container)
plt.tight_layout()
plt.show()
```

**INTERPRETATION**:
Instagram, TikTok, and WhatsApp show the highest counts of students saying social media affects their academics. Facebook and LinkedIn opposes this trend, with more "No" responses than "Yes," suggesting they're seen as less academically disruptive.

**Comparing the most used social media platform across the top 10 countries :**

```python
top10 = df['Country'].value_counts().head(10).index
plt.figure(figsize=(12,6))
ax = sns.countplot(data=df[df['Country'].isin(top10)],x='Country',hue='Most_Used_Platform',palette='pastel',edgecolor='black')
plt.title('Most Used Platform by Country (Top 10 Countries)')
plt.xlabel('Country')
plt.ylabel('Number of Students')
plt.xticks(rotation=45)
plt.legend(loc='upper right')
for container in ax.containers:
    ax.bar_label(container, fontsize=8)
plt.tight_layout()
plt.show()
```

**INTERPRETATION**:
WhatsApp dominates in France and Switzerland, while Instagram leads in Denmark, Turkey, and USA, showing platform preference varies heavily by country rather than one platform being universal. TikTok is notably strong in Canada and Spain, suggesting regional/cultural pull rather than a single global favorite among students.

# CONCLUSION:
The dataset analyzes students' **social media usage, addiction, sleep patterns, and mental health** across different age groups, academic levels, and countries. Most students are **20 years old**, followed by **19** and **21**. Female students show slightly higher social media addiction than males, although the difference is minimal. Addiction is highest among **19-year-olds** and generally decreases with age, with **24-year-olds** having the lowest addiction scores.
Among academic levels, **undergraduate students exhibit the highest addiction**, while **high school students** have the lowest. **India** records the highest addiction scores, followed by the **USA** and **Canada**. **Instagram** is the most popular platform (about **36%** of users), followed by **TikTok, Facebook, and WhatsApp**.
The analysis also indicates that **higher social media addiction is associated with poorer mental health and fewer hours of sleep**. In contrast, students with **better mental health tend to sleep longer and experience fewer social media-related conflicts**. Overall, the findings suggest a strong relationship between excessive social media use and reduced student well-being.
